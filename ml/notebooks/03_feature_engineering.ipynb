# 03 - Feature Engineering

Features are produced by `data_pipeline/src/features/aggregator.py`
(window aggregation incl. port_scan_score heuristic and PCAP-only features).
This notebook reviews which features survived and their distributions.

```python
import json, numpy as np
schema = json.load(open('ml/artifacts/feature_schema.json'))
names = schema['feature_names']; print(len(names), 'features'); names
```

```python
import matplotlib.pyplot as plt
t = np.load('ml/artifacts/tensors.npz') if __import__('pathlib').Path('ml/artifacts/tensors.npz').exists() else None
if t is not None:
    flat = t['X'].reshape(-1, len(names))
    plt.boxplot(flat, labels=names, vert=False); plt.title('normalized features'); plt.show()
```
