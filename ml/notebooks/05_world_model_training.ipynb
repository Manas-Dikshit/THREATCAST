# 05 - World Model Training

Trains the multi-head Temporal Transformer for **future-state prediction**
(the world-model objective), plus auxiliary heads.
CLI equivalent: `python -m ml.src.training.train --dataset-dir <export>`

```python
from ml.src.utils.config import WorldModelConfig
cfg = WorldModelConfig.load()
cfg.training.epochs = 10        # notebook-scale run
cfg.model.d_model, cfg.model.n_layers = 64, 2
cfg
```

```python
from ml.src.training.trainer import Trainer
trainer = Trainer(cfg, 'ml/artifacts')
result = trainer.train('ml/artifacts')   # dataset dir with tensors.npz
result['history'][-1]
```

Loss curve - confirm the world-model term (`train_next_state`) decreases:

```python
import matplotlib.pyplot as plt
ns = [r['train_next_state'] for r in result['history']]
plt.plot(ns); plt.xlabel('epoch'); plt.ylabel('next-state MSE'); plt.show()
```
